```python
# Importando Funções Pyspark
from pyspark.sql.functions import *

#=======================#
# Objetivo das Análises #
#=======================#

#1. Quais os estados brasileiros registraram maior número de acidentes.
#2. Quais meses do ano tem mais ocorrências de acidentes.
#3. Quais tipos de acidentes mais frequentes.
```

```python
# Carregando Tabela Silver
df_gold = spark.table("projeto_profissional_bigdata.datatran_2018.silver_datatran18")
```

```python
# Visualizando tabela gold.

df_gold.printSchema()
df_gold.show()
display(df_gold)
```

```python
# Pergunta `1` = Estado com maior número de acidentes.

gold_acidente_por_estado = (
    df_gold.groupBy("uf")
    .agg(count("*").alias("Total_Acidentes"))
    .orderBy(col("Total_Acidentes").desc())
)

print ("Estado com Maior Número de Acidentes:")
gold_acidente_por_estado.show(27)
```

```python
# Salvando Tabela Gold

gold_acidente_por_estado.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("projeto_profissional_bigdata.datatran_2018.gold_acidente_por_estado")
```

```python
%sql
SELECT * FROM projeto_profissional_bigdata.datatran_2018.gold_acidente_por_estado
```

```python
# Pergunta `2` = Quais meses do ano tem mais ocorrência.

gold_acidente_por_mes = (
    df_gold.groupBy("mes_acidente", "nome_mes")
    .agg(count("*").alias("total_acidentes"))
    .orderBy(col("mes_acidente"))
)

print ("Quantidade de acidentes por mês:")
gold_acidente_por_mes.show(12)
     
```

```python
# Salvando Tabela Gold

gold_acidente_por_mes.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("projeto_profissional_bigdata.datatran_2018.gold_acidente_por_mes")
     
```

```python
%sql
SELECT * FROM projeto_profissional_bigdata.datatran_2018.gold_acidente_por_mes
     
```

```python
df_gold.printSchema()
```

```python
#3. Quais tipos de acidentes mais frequentes.

gold_acidente_por_tipo = (
    df_gold.groupBy("tipo_acidente")
    .agg(count("*").alias("total_acidentes"))
    .orderBy(col("total_acidentes").desc())
)

print ("Acidentes por tipo:")
gold_acidente_por_tipo.show(30, truncate=False)
     
```

```python
# Salvando Tabela Gold

gold_acidente_por_tipo.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("projeto_profissional_bigdata.datatran_2018.gold_acidente_por_tipo")
     
```
